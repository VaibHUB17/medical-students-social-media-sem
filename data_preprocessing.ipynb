```python
# Performing sentiment analysis over the answers of Q19 as it depicts the comments of the respondents
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer
nltk.download('vader_lexicon')
sia = SentimentIntensityAnalyzer()
def categorize_sentiment(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    else:
        return "Neutral"
df['Q19'] = df['Q19'].apply(lambda x: categorize_sentiment(sia.polarity_scores(str(x))['compound']))
df['Q19'] = lb.fit_transform(df[['Q19']]).astype(str)
df['Q19'].unique()
```

```python
import semopy as sem
df_clean = df.apply(pd.to_numeric, errors='coerce')  # force numeric
df_clean = df_clean.dropna()  # drop rows with NaN
```

```python
# order of the variables matter in semopy
model_desc =''' 
# Measurement model
age_interest        =~ Q2 + Q8
security_concerns   =~ Q15 + Q16 + Q17
unwanted_use        =~ Q6 + Q7 + Q14 + Q15
academic_performance =~ Q9 + Q10
motivation          =~ Q10 + Q11 + Q12 + Q13
interest            =~ Q13 + Q18
usage               =~ Q6 + Q7

# Structural model
academic_performance ~ usage + age_interest + motivation
motivation ~ age_interest + usage
security_concerns ~ unwanted_use
interest ~ motivation + academic_performance
'''
model = sem.Model(model_desc)
res = model.fit(df_clean)
print(res)
```

```python
est= model.inspect()
print(est)
```

```python
import tabulate
stat=sem.calc_stats(model)
print(tabulate.tabulate(stat,headers='keys',tablefmt='fancy_grid'))
```
